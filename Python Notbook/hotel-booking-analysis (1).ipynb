## Handling Missing Values (`company` & `agent`)

Instead of applying a generic imputation (e.g., filling all missing values with the mode or removing the rows), we first analyzed the business context behind the missing values to determine whether they represent missing information or meaningful business scenarios.

### Company Column

After analyzing the relationship between `company` and `market_segment`, we observed the following:

- Most bookings from **Direct**, **Online TA**, **Offline TA/TO**, **Groups**, **Complementary**, and **Aviation** do not have an associated company ID.
- For these booking types, a missing value most likely indicates that the reservation was **not made on behalf of a company**, rather than a missing record.
- In contrast, approximately **84%** of **Corporate** bookings have a valid company ID. Therefore, when a Corporate booking has a missing company value, it is more appropriate to classify it as **"Unknown Company"**, indicating that the company information is missing rather than nonexistent.

**Imputation Strategy**

- `Corporate + Missing Company` → **Unknown Company**
- `All Other Market Segments + Missing Company` → **No Company**

---

### Agent Column

The analysis of `agent` together with `market_segment` and `distribution_channel` revealed two different scenarios:

- For bookings where both `market_segment` and `distribution_channel` are **Direct**, a missing agent value logically means that the customer booked **directly with the hotel**, without using a travel agent.
- For bookings coming through channels such as **TA/TO** or **Corporate**, a missing agent value is more likely to represent an **unknown or unrecorded travel agent**, rather than the absence of an agent.

**Imputation Strategy**

- `Direct + Direct + Missing Agent` → **Direct Booking**
- `All Other Missing Agent Values` → **Unknown Agent**

---

### Why This Approach?

This business-driven imputation preserves the meaning of the data instead of replacing missing values with arbitrary values such as the mode or deleting observations.

As a result:

- Missing values are transformed into meaningful business categories.
- The dataset becomes easier to analyze and visualize.
- Business insights remain accurate and interpretable without introducing misleading information.

### 0.1 Analysis configuration and reusable helpers

Defining the plotting theme, colour palette and a small set of helper functions once keeps the
rest of the notebook short and the visual language consistent. Each helper is used many times
below:

| Helper | Purpose |
|---|---|
| `iqr_bounds` / `iqr_screen` | compute Tukey fences and report what they *would* flag — a screen, never a decision |
| `describe_plus` | `describe()` plus skewness, kurtosis and mode, which is what actually drives transformation choices |
| `profile_numeric` | histogram + KDE + boxplot for one numeric column |
| `cancel_rate_by` | cancellation rate, volume and share for every level of a categorical column |
| `bar_with_labels` | labelled bar chart with an optional baseline reference line |

```python
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    "figure.dpi": 110, "savefig.dpi": 110, "font.size": 10,
    "axes.titlesize": 12, "axes.titleweight": "bold", "axes.labelsize": 10,
    "axes.edgecolor": "#666666", "grid.color": "#E5E5E5", "figure.autolayout": False,
})

# consistent palette used for every chart in this notebook
NAVY, TEAL, CORAL, GOLD, GREY = "#1F3B57", "#2A9D8F", "#E76F51", "#E9C46A", "#8D99AE"
SEQ = "mako"

MONTH_ORDER = ['January','February','March','April','May','June',
               'July','August','September','October','November','December']
MONTH_NUM = {m: i + 1 for i, m in enumerate(MONTH_ORDER)}


def iqr_bounds(s, k=1.5):
    '''Classic Tukey fence for a numeric Series.'''
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    return q1 - k * iqr, q3 + k * iqr


def iqr_screen(df, cols, k=1.5):
    '''Screen (never decide) -- how many rows each Tukey fence would flag.'''
    rows = []
    for c in cols:
        lo, hi = iqr_bounds(df[c], k)
        flagged = ((df[c] < lo) | (df[c] > hi)).sum()
        rows.append({
            "column": c, "Q1": df[c].quantile(.25), "Q3": df[c].quantile(.75),
            "lower_fence": lo, "upper_fence": hi, "min": df[c].min(), "max": df[c].max(),
            "flagged": flagged, "flagged_%": round(100 * flagged / len(df), 2),
            "skew": round(df[c].skew(), 2), "zeros_%": round(100 * (df[c] == 0).mean(), 1),
        })
    return pd.DataFrame(rows).sort_values("flagged_%", ascending=False).reset_index(drop=True)


def describe_plus(df, cols):
    '''describe() plus the shape statistics that decide transformation choices.'''
    d = df[cols].describe(percentiles=[.01, .25, .5, .75, .95, .99]).T
    d["skew"] = df[cols].skew()
    d["kurtosis"] = df[cols].kurtosis()
    d["mode"] = [df[c].mode().iloc[0] for c in cols]
    return d.round(2)


def profile_numeric(df, col, color=NAVY, log_x=False, clip_q=None, title=None):
    '''Histogram + KDE + boxplot for one numeric column.'''
    s = df[col].dropna()
    shown = s[s <= s.quantile(clip_q)] if clip_q else s
    fig, ax = plt.subplots(1, 3, figsize=(15, 3.6))
    sns.histplot(shown, bins=50, color=color, ax=ax[0], edgecolor="white", linewidth=.3)
    ax[0].set_title(f"{col} - histogram"); ax[0].set_xlabel(col); ax[0].set_ylabel("Bookings")
    sns.kdeplot(shown, color=color, fill=True, alpha=.30, ax=ax[1], linewidth=1.6)
    ax[1].set_title(f"{col} - density"); ax[1].set_xlabel(col)
    sns.boxplot(x=s, color=color, ax=ax[2], fliersize=2, width=.45)
    ax[2].set_title(f"{col} - boxplot (full range)"); ax[2].set_xlabel(col)
    if log_x:
        ax[2].set_xscale("symlog")
    fig.suptitle(title or f"Distribution profile - {col}", fontsize=13, fontweight="bold", y=1.04)
    plt.tight_layout(); plt.show()
    print(f"mean={s.mean():.2f} | median={s.median():.2f} | mode={s.mode().iloc[0]} | "
          f"std={s.std():.2f} | skew={s.skew():.2f} | kurtosis={s.kurtosis():.2f}")


def cancel_rate_by(df, col, min_n=1, sort=True):
    '''Cancellation rate + volume for every level of a categorical column.'''
    g = df.groupby(col, observed=True)['is_canceled'].agg(bookings='size', cancel_rate='mean')
    g = g[g['bookings'] >= min_n]
    g['cancel_%'] = (100 * g['cancel_rate']).round(2)
    g['share_%'] = (100 * g['bookings'] / len(df)).round(2)
    g = g.drop(columns='cancel_rate')
    return g.sort_values('cancel_%', ascending=False) if sort else g


def bar_with_labels(ax, x, y, color=NAVY, fmt="{:.1f}%", rotate=0, ylabel="", title="", baseline=None):
    '''Reusable labelled bar chart -- used for every rate comparison in this notebook.'''
    bars = ax.bar([str(v) for v in x], y, color=color, edgecolor="white", linewidth=.6)
    ax.bar_label(bars, labels=[fmt.format(v) for v in y], padding=2, fontsize=8.5)
    ax.set_ylabel(ylabel); ax.set_title(title)
    ax.set_ylim(0, max(y) * 1.20 if len(y) else 1)
    if baseline is not None:
        ax.axhline(baseline, ls="--", lw=1.2, color=CORAL, zorder=0)
        ax.text(0.995, baseline, f" overall {baseline:.1f}%", color=CORAL, fontsize=8,
                va="bottom", ha="right", transform=ax.get_yaxis_transform())
    if rotate:
        ax.tick_params(axis="x", rotation=rotate)
        for t in ax.get_xticklabels():
            t.set_ha("right")
    return ax


OVERALL_CANCEL = 100 * data['is_canceled'].mean()
NUM_COLS = data.select_dtypes(include=np.number).columns.drop('is_canceled').tolist()
CAT_COLS = ['hotel','meal','market_segment','distribution_channel','deposit_type',
            'customer_type','reserved_room_type','assigned_room_type','reservation_status']

print(f"Overall cancellation rate: {OVERALL_CANCEL:.2f}%")
print(f"{len(NUM_COLS)} numeric feature columns | {len(CAT_COLS)} core categorical columns")
```

# Step 1 — Outlier Analysis

## 1.0 Method: screen statistically, decide commercially

An outlier is only a problem if it is *wrong*. In a hotel booking system most extreme values
are not errors — they are the long tail of real commercial behaviour (a suite in peak August,
a 60-night extended stay, a group block booked two years out). So the Tukey fence

$$\text{fence} = \left[\, Q_1 - 1.5 \cdot IQR,\ \ Q_3 + 1.5 \cdot IQR \,\right]$$

is used here **only as a screen**: it tells us where to look. Every column is then classified
into one of three buckets, and only the first justifies destructive treatment:

| Classification | Meaning | Typical treatment |
|---|---|---|
| **Data entry error** | Physically impossible or internally contradictory | Remove or correct |
| **Statistical outlier only** | Rare but commercially valid | Keep |
| **Legitimate business value** | Extreme *because* the business works that way | Keep, sometimes flag |

The screen runs first so we can see how badly a blind IQR rule would perform.

```python
screen = iqr_screen(data, NUM_COLS)
print(screen.to_string(index=False))
print(f"\nTotal rows: {len(data):,}")
print("Columns where the Tukey fence flags >5% of the data:",
      screen.loc[screen['flagged_%'] > 5, 'column'].tolist())
```

### 1.1 Why a blind IQR rule would destroy this dataset

The screen exposes two structural problems that make automatic IQR removal unusable here.

**Problem 1 — zero-width IQR on discrete columns.** For `adults`, $Q_1 = Q_3 = 2$, so
$IQR = 0$ and the fence collapses to the single point $[2, 2]$. Every solo traveller and
every party of three becomes an "outlier": **22,899 rows (26.2%)** would be deleted, including
16,503 perfectly ordinary single-guest bookings. The same collapse happens on `booking_changes`,
`children` and `required_car_parking_spaces`.

**Problem 2 — zero-inflation.** Six columns are more than 90% zeros
(`previous_cancellations` 98%, `days_in_waiting_list` 99%, `babies` 99%,
`required_car_parking_spaces` 92%, `children` 90%). These are *event counters*: zero means
"the event did not happen", which is the normal case. Any non-zero value sits outside the fence
by construction, so the IQR rule would delete precisely the rows that carry the signal.

Applied naively across all columns, the union of these fences would discard well over a third
of the dataset — and would specifically remove `previous_cancellations`, the single strongest
behavioural predictor of cancellation in the data (67.9% vs 26.7%).

The rest of Step 1 therefore works column by column.

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))

sub = screen.head(10)
bar_with_labels(ax[0], sub['column'], sub['flagged_%'].values, color=GREY, fmt="{:.1f}%",
                rotate=35, ylabel="% of rows flagged", title="Rows the Tukey fence would delete")

zi = pd.DataFrame({"column": NUM_COLS, "zeros_%": [100 * (data[c] == 0).mean() for c in NUM_COLS]})
zi = zi.sort_values("zeros_%", ascending=False).head(10)
bar_with_labels(ax[1], zi['column'], zi['zeros_%'].values, color=TEAL, fmt="{:.0f}%",
                rotate=35, ylabel="% of rows equal to zero", title="Zero-inflation by column")

fig.suptitle("Why the IQR rule cannot be applied blindly to this dataset",
             fontsize=13, fontweight="bold", y=1.03)
plt.tight_layout(); plt.show()

print("adults  -> Q1 = Q3 = {} => IQR = 0".format(data['adults'].quantile(.25)))
print(data['adults'].value_counts().sort_index().head(4).to_string())
print("\nA zero-width IQR makes every value except 2 an 'outlier':",
      "{:,} rows ({:.1f}%)".format(
          int(((data['adults'] < 2) | (data['adults'] > 2)).sum()),
          100 * ((data['adults'] < 2) | (data['adults'] > 2)).mean()))
```

#### Business meaning and decision — `adr`

The distribution is dominated by a single record. Four distinct findings, four different
treatments:

**a) ADR = 5,400 — one record. → REMOVE.**
The next highest rate in 87,396 bookings is **510**, so this value is more than ten times the
observed maximum, for a *one-night stay in a room type A* (the cheapest category). No rate card
supports it. It is a decimal/keying error, and it alone is responsible for the pathological
shape statistics: skewness **10.92** and kurtosis **981.6**. The booking was cancelled, so it
never generated revenue and removing it costs no commercial information.

**b) ADR = −6.38 — one record. → FLOOR AT 0.**
A negative room rate is impossible. The rest of the record is coherent (a Groups
Transient-Party booking, 10 nights, checked out), so the correct treatment is to correct the
impossible field rather than delete a valid stay. Most likely a credit or rate adjustment
posted against the booking.

**c) ADR = 0 — 1,778 records (2.03%). → KEEP.**
These are *not* errors. **91% of the entire Complementary segment** is zero-rated, which is
exactly what "complementary" means: comps, staff stays, loyalty redemptions and contra deals.
They also behave differently and sensibly — a **10.4%** cancellation rate versus 27.5% overall,
because a free room is rarely cancelled. Deleting them would erase a real customer category and
bias occupancy analysis. They are kept and flagged later as `is_zero_rated`.

**d) 2,490 records above the fence (227). → KEEP.**
The fence sits at 227 while p99 is **261** and p99.9 is **335**. The values between 227 and 510
form a smooth continuum — family suites (room types G/H) with 2 adults + 2 children in peak
season. This is the hotel's most valuable business, not noise. Capping it would systematically
understate high-season revenue.

#### Business meaning and decision — party size

**a) `adults` between 10 and 55 — 13 records. → KEEP + FLAG.**
The IQR rule screams "error" here, but the evidence says otherwise. All 13 records share an
identical signature: **every one** is `customer_type = Group`, **every one** has `ADR = 0`,
all are Resort Hotel, all Portuguese, all booked 304–361 days ahead, and all were cancelled.
This is not 13 independent typos — it is a **group allocation block**, where a tour operator
holds N beds on one placeholder record at zero rate before individual bookings are attached.
That is standard hotel inventory practice. They are kept and flagged as `is_group_block` so
per-booking occupancy averages can exclude them on demand.

**b) Zero total guests — 166 records. → REMOVE.**
A booking with no adults, no children and no babies describes nobody staying in the room. It
cannot be a stay under any interpretation, and **135 of the 166** also carry zero ADR, i.e. no
revenue either. These are administrative shell records. This is the one party-size group that
genuinely warrants deletion.

**c) Children/babies but no adults — 219 records. → KEEP + FLAG.**
Again a clear signature rather than random noise: **all 219 are City Hotel**, and **205 of them
(94%) are room type `B`**, carrying a real median ADR of 87. Unaccompanied minors are not a real
hotel product; the consistent room type plus a live rate points to a **child-supplement line
attached to a parent booking**. Deleting them would remove real revenue, and imputing a phantom
adult would invent occupancy. They are kept and flagged as `is_child_only_record`.

**d) 9–10 babies / 10 children — 3 records. → CAP.**
`babies = 10` against 2 adults and `babies = 9` against 1 adult are physically implausible, as is
`children = 10` in a single room. Unlike the group blocks these have no supporting pattern — three

### 1.8 Applying the treatment

Removals are deliberately conservative: **758 rows, 0.87% of the dataset**, and every one of
them is a record that cannot describe a real stay (no guests, no nights, or an impossible rate).
Nothing is removed for being merely extreme.

The result is stored in `data_clean` so the pre-treatment frame remains available for comparison.

```python
data = data.drop(columns=[c for c in data.columns if c.startswith('_')])
before_rows, before_skew, before_kurt = len(data), data['adr'].skew(), data['adr'].kurtosis()

total_guests = data['adults'] + data['children'] + data['babies']
total_nights = data['stays_in_week_nights'] + data['stays_in_weekend_nights']

drop_no_guest  = total_guests == 0
drop_no_night  = total_nights == 0
drop_adr_error = data['adr'] > 1000
to_drop = drop_no_guest | drop_no_night | drop_adr_error

print("Rows removed")
print("  zero total guests              : {:>5,}".format(int(drop_no_guest.sum())))
print("  zero total nights              : {:>5,}".format(int(drop_no_night.sum())))
print("  impossible ADR (> 1000)        : {:>5,}".format(int(drop_adr_error.sum())))
print("  overlap between the above      : {:>5,}".format(
    int(drop_no_guest.sum() + drop_no_night.sum() + drop_adr_error.sum() - to_drop.sum())))
print("  ------------------------------------")
print("  unique rows removed            : {:>5,} ({:.2f}% of the dataset)".format(
    int(to_drop.sum()), 100 * to_drop.mean()))

data_clean = data.loc[~to_drop].copy()

# corrective caps -- fix the impossible field, keep the valid stay
data_clean['adr']    = data_clean['adr'].clip(lower=0)
data_clean['babies'] = data_clean['babies'].clip(upper=2)
data_clean['children'] = data_clean['children'].clip(upper=3)
data_clean['required_car_parking_spaces'] = data_clean['required_car_parking_spaces'].clip(upper=3)

data_clean = data_clean.reset_index(drop=True)

print("\nShape: {} -> {}".format((before_rows, data.shape[1]), data_clean.shape))
print("ADR skew    : {:>8.2f} -> {:>5.2f}".format(before_skew, data_clean['adr'].skew()))
print("ADR kurtosis: {:>8.2f} -> {:>5.2f}".format(before_kurt, data_clean['adr'].kurtosis()))
print("Cancellation rate: {:.2%} -> {:.2%}  (target balance preserved)".format(
    data['is_canceled'].mean(), data_clean['is_canceled'].mean()))
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 3.6))
sns.histplot(data['adr'], bins=60, color=GREY, ax=ax[0], edgecolor="white", linewidth=.3)
ax[0].set_title("Before - skew {:.2f}".format(before_skew)); ax[0].set_xlabel("ADR")
sns.histplot(data_clean['adr'], bins=60, color=NAVY, ax=ax[1], edgecolor="white", linewidth=.3)
ax[1].set_title("After  - skew {:.2f}".format(data_clean['adr'].skew())); ax[1].set_xlabel("ADR")
sns.histplot(np.log1p(data_clean['adr']), bins=60, color=CORAL, ax=ax[2], edgecolor="white", linewidth=.3)
ax[2].set_title("log1p(ADR) - skew {:.2f} (worse)".format(np.log1p(data_clean['adr']).skew()))
ax[2].set_xlabel("log1p(ADR)")
fig.suptitle("Removing one erroneous record fixed the distribution; a log transform would break it",
             fontsize=13, fontweight="bold", y=1.04)
plt.tight_layout(); plt.show()

print("Zero-ADR bookings retained: {:,}".format(int((data_clean['adr'] == 0).sum())))
print("Because log1p compresses the right tail while the 1,053 legitimate zeros stay pinned at 0,")
print("the transform converts mild right-skew (+0.93) into severe left-skew ({:.2f}). No transform applied."
      .format(np.log1p(data_clean['adr']).skew()))
```

#### Was a transformation needed? No — and this is why it matters

| Column | Skew before | Skew after | Treatment | Transform? |
|---|---|---|---|---|
| `adr` | **10.92** | **0.93** | removed 1 erroneous record | **No** |
| `lead_time` | 1.43 | 1.42 | none | No |
| `previous_cancellations` | 34.32 | 34.23 | none | No — zero-inflated by design |

ADR's kurtosis fell from **981.6 to 1.58** because *one* wrong record was deleted, not because a
transform was applied. This is the practical argument for diagnosing before transforming: had a
log been applied first, it would have "worked" (the skew number would have improved) while leaving
the 5,400 error sitting in the data.

And the log transform is actively harmful here. Because 1,053 legitimate zero-rated bookings stay
pinned at $\log(1+0) = 0$ while the right tail compresses, `log1p(adr)` turns mild right-skew of
**+0.93** into severe *left*-skew of **−3.49**. The remaining skew is mild and financial variables
are naturally right-skewed, so `adr` is left on its original, interpretable euro scale.

The zero-inflated counters are also left untransformed: no monotone transform fixes a 98% zero
mass, and the correct treatment for them is the binary/categorical encoding built in Step 3.

```python
def dataset_overview(df):
    '''One-call structural audit: shape, memory, duplicates, missingness, cardinality.'''
    print("SHAPE: {:,} rows x {} columns".format(*df.shape))
    print("MEMORY: {:.2f} MB (deep)".format(df.memory_usage(deep=True).sum() / 1024**2))
    print("DUPLICATES: {}".format(df.duplicated().sum()))
    print("MISSING CELLS: {}\n".format(int(df.isnull().sum().sum())))
    ov = pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "non_null": df.notnull().sum(),
        "missing": df.isnull().sum(),
        "unique": df.nunique(),
        "unique_%": (100 * df.nunique() / len(df)).round(2),
        "most_frequent": [df[c].value_counts().index[0] if df[c].nunique() else None for c in df.columns],
        "freq_%": [round(100 * df[c].value_counts().iloc[0] / len(df), 2) if df[c].nunique() else None
                   for c in df.columns],
    })
    return ov


overview = dataset_overview(data_clean)
print(overview.to_string())
```

tails (`L` and `P` have single-digit counts and their rates are not interpretable).

**A data-quality note on `meal`:** the 485 `Undefined` records are worth a second look. The common
advice is to merge `Undefined` into `SC` (Self Catering), but the evidence here contradicts that:
all 485 are **Resort Hotel**, whereas `SC` is 99% **City Hotel**, and they cancel at **16.9%** versus
**35.7%** for `SC`. They behave like a distinct product, so merging them would blur two different
things. `Undefined` is left as its own level.

**A second data-quality note on `country`:** both `CN` (1,092) and `CHN` (815) appear, i.e. the ISO-2
and ISO-3 codes for China coexist. Their cancellation rates differ sharply (20.6% vs 48.1%), which
suggests they arrive from two different source systems rather than being a simple duplicate label.
Flagged rather than merged — this needs a source-system answer from the business before either
combining or separating them.

## 2.3 Bivariate Analysis

The central business question is *what predicts a cancellation*. Every panel below compares the
cancellation rate of each level against the **27.7% baseline** (dashed line); red bars are worse
than baseline, teal better. Volumes (`n=`) are printed inside each bar so small, unreliable

#### Interpretation — the counter-intuitive deposit finding

**`Non Refund` bookings cancel at 94.7%** — the single most surprising number in the dataset. A
non-refundable deposit is supposed to *prevent* cancellation, and here it accompanies near-certain
cancellation. Two mechanisms explain it, and the distinction matters:

1. **The policy is applied as a reaction to risk, not as a deterrent.** Of the 1,037 `Non Refund`
   bookings, **982 are Portuguese** and mean lead time is **211 days** versus 80 overall. Revenue
   management is imposing pre-payment terms on exactly the profile it already distrusts. The deposit
   is a *marker* of risk, not a cure for it.
2. **A cancelled non-refundable booking is not a lost sale.** The guest forfeits the payment, so the
   revenue is retained. §4.1 quantifies this: it converts €237k of the apparent loss into retained
   revenue.

The practical implication is that `deposit_type` must not be read causally. "Require deposits to cut
cancellations" does not follow from this table — the correct reading is that the current policy
successfully *identifies* high-risk bookings and monetises them, but does nothing to convert them
into stays.

**Reliable channels:** `Corporate` (12.2%), `Direct` (14.9%) and `Offline TA/TO` (15.0%) all run at
roughly half the baseline. **Online TA (35.6%)** is the problem channel and it is also the largest.

**Special requests are the cleanest protective signal in the data** — a perfect monotone gradient
from **33.5%** at zero requests to **5.6%** at five. A guest who asks for a high floor and a late
check-in has mentally committed to the trip. Because requests are recorded at booking time, this is
usable for prediction (unlike parking — see §2.6).

```python
data_clean['lead_time_bin'] = pd.cut(
    data_clean['lead_time'], bins=[-1, 0, 7, 30, 90, 180, 365, 10_000],
    labels=['0 (same day)','1-7','8-30','31-90','91-180','181-365','365+'])

g = cancel_rate_by(data_clean, 'lead_time_bin', sort=False)

fig, ax = plt.subplots(1, 2, figsize=(14, 4.2))
bar_with_labels(ax[0], g.index, g['cancel_%'].values, color=NAVY, fmt="{:.1f}%",
                ylabel="Cancellation rate (%)",
                title="Cancellation rate rises monotonically with lead time",
                baseline=BASE, rotate=25)
ax[0].set_xlabel("Days between booking and arrival")

for lbl, sub, c in [("Kept", data_clean[data_clean.is_canceled == 0], TEAL),
                    ("Cancelled", data_clean[data_clean.is_canceled == 1], CORAL)]:
    sns.kdeplot(sub['lead_time'], ax=ax[1], label=lbl, fill=True, alpha=.28, color=c, linewidth=1.6)
ax[1].set_title("Lead-time density by outcome"); ax[1].set_xlabel("Lead time (days)")
ax[1].set_xlim(0, 500); ax[1].legend(frameon=False)
plt.tight_layout(); plt.show()

print(g.to_string())
print("\nMedian lead time - cancelled: {:.0f} days | kept: {:.0f} days".format(
    data_clean.loc[data_clean.is_canceled == 1, 'lead_time'].median(),
    data_clean.loc[data_clean.is_canceled == 0, 'lead_time'].median()))
```

#### Interpretation — lead time is the strongest continuous driver

The gradient is monotone and steep: **6.0%** for same-day bookings rising to **41.0%** beyond a
year — a **6.8x** spread. Median lead time is **80 days for cancelled** bookings versus **38 days
for kept** ones.

The mechanism is intent decay. A booking made months ahead is a provisional plan competing with
alternatives that have not been offered yet; a booking made this week is a decision. The KDE panel
shows kept bookings crowded against the left axis while cancellations spread across the horizon.

The business break is between **1–7 days (9.7%)** and **8–30 days (25.5%)** — risk nearly triples in
that step, which is where an overbooking model should change its assumptions.

#### Interpretation — the two hotels are different businesses

Comparing headline ADR (City **111.66** vs Resort **100.12**) suggests the City Hotel is the
stronger performer. The monthly view shows that conclusion is wrong, or at least incomplete.

**The two properties have opposite seasonal shapes.** The Resort swings from **49.92** in January to
**188.90** in August — a **3.8x** annual range. The City Hotel runs between 85.84 and 128.36, a
**1.5x** range. The Resort earns its year in roughly ten weeks; the City Hotel earns steadily.

**They are also priced against different demand.** The Resort out-prices the City Hotel in June–September
and is far cheaper the rest of the year. So "which hotel is better" is the wrong question — the
Resort is a yield-management business, the City Hotel a volume-and-occupancy business, and they need
separate pricing and staffing models. Realised revenue per surviving booking confirms this: **422**
at the Resort versus **328** at the City Hotel.

**Demand and risk move together.** August is simultaneously the highest-volume month (11,194
bookings), the highest-rate month, and among the highest-cancelling (32.2%). Peak season carries peak

### Reserved vs assigned room — and a warning about target leakage

The comparison of `reserved_room_type` with `assigned_room_type` is usually presented as an
upgrade/downgrade analysis. It also surfaces the most important modelling caveat in this dataset.

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4.2))

ct = pd.crosstab(data_clean.reserved_room_type, data_clean.assigned_room_type)
sns.heatmap(ct, annot=True, fmt=",d", cmap=SEQ, ax=ax[0], cbar_kws={'label': 'Bookings'},
            annot_kws={"size": 7}, linewidths=.4, linecolor='white')
ax[0].set_title("Reserved vs assigned room type"); ax[0].set_xlabel("Assigned"); ax[0].set_ylabel("Reserved")

bar_with_labels(ax[1], ['reservation_status\n(perfect proxy)','parking > 0','room changed'],
                [100.0,
                 100 * data_clean.loc[data_clean.required_car_parking_spaces > 0, 'is_canceled'].mean(),
                 100 * data_clean.loc[data_clean.room_changed == 1, 'is_canceled'].mean()],
                color=[CORAL, GOLD, GOLD], fmt="{:.2f}%",
                ylabel="Cancellation rate (%)", title="Target-leaking fields", baseline=BASE)

plt.tight_layout(); plt.show()
```

#### Interpretation — three fields leak the outcome

Room changes affect **14.6%** of bookings and are concentrated on the diagonal-adjacent cells
(A→C, A→D, D→E): mostly operational reallocation and upgrades within a tier, not downgrades.
The Resort reallocates far more (20.2%) than the City Hotel (11.1%), consistent with a
smaller, more heterogeneous room stock.

But the cancellation figures reveal something that has nothing to do with room preference:

| Field | Cancellation rate | Why it leaks |
|---|---|---|
| `reservation_status` | 100% / 0% | `Canceled` + `No-Show` **is** `is_canceled = 1`, exactly. A perfect restatement of the target. |
| `required_car_parking_spaces > 0` | **0.00%** (0 of 7,290) | Not a single cancellation in 7,290 bookings. Parking is allocated **on arrival**, so a non-zero value proves the guest arrived. |
| `room_changed = 1` | **4.8%** vs 31.6% | Rooms are assigned **at check-in**. Only 1.9% of cancelled bookings show a change, versus 19.2% of check-outs. |

None of these is a *predictor* of cancellation; all three are *consequences* of the guest turning up.
The r = −0.187 correlation between parking and cancellation in §2.6 is entirely an artifact of this.

**Practical consequence:** these fields must be excluded from any predictive model. Including them
would produce near-perfect validation scores that collapse in production, because at prediction time
(when the booking is made) none of them is known. They remain useful for *descriptive* analysis of
what happened, which is how they are used above. The exclusion list is formalised in §3.6.

#### Interpretation — the domestic-market interaction is the hidden pattern

**Guest origin dominates everything else, and it is invisible in single-variable analysis.**
Portugal is 31% of bookings and cancels at 36.4% versus 23.8% internationally. But the *interaction*
is far stronger than that average:

| Market segment | International | Portugal | Ratio |
|---|---|---|---|
| **Groups** | **1.6%** | **58.8%** | **37x** |
| **Offline TA/TO** | **0.5%** | **44.4%** | **89x** |
| Corporate | 2.7% | 15.1% | 5.6x |
| Direct | 10.8% | 19.6% | 1.8x |
| Online TA | 33.2% | 44.4% | 1.3x |

Domestic Offline TA/TO bookings cancel at **44.4%** while the identical channel sells to
international guests at **0.5%** — the same product, an 89-fold difference in reliability. Averaging
these together produces the misleading "Offline TA/TO is a safe channel at 15.0%" conclusion from
§2.3. The channel is not safe; it is safe *internationally* and among the worst domestically.

The likely mechanism is speculative domestic holding: local guests and local operators block rooms
with no deposit and no travel cost sunk, then decide later. International guests have flights booked.

**This single interaction is the most actionable finding in the analysis** — it identifies a specific,
sizeable book of business (roughly 6,700 domestic Groups + Offline TA/TO bookings) where a deposit
requirement would have a large effect and would not touch the reliable international base.

**Deposit x lead time** confirms §2.3: `Non Refund` sits at 94–100% across every horizon beyond a

**Rate x segment:** `Online TA` + `Transient` yields the highest mean ADR (**119**) and `Direct` +
`Transient` is close behind (**118**), while `Corporate` runs at **50–83**. The premium segments are
also the volatile ones; the stable segments are cheap. There is no quadrant that is both.

The effect is largest exactly where it is most useful: in the 181–365 day band, requests cut the rate
from **46.0%** to **31.7%**. This makes "number of special requests at booking" a genuinely useful
early-warning field, and it suggests a cheap operational intervention — proactively soliciting
preferences from long-lead bookers converts them into engaged, committed guests.

(The 365+ row shows a sharp drop to 9.2%/11.9%, but on only ~150 bookings split across the columns;
treat that specific cell as noise.)

#### Interpretation — weak linear correlations, strong categorical effects

**Every numeric correlation with the target is below |0.19|.** Read naively this says "nothing
predicts cancellation", which flatly contradicts the 94.7% and 58.8% rates found above. Three
reasons, and each is a general lesson:

1. **The real drivers are categorical.** `deposit_type`, `market_segment` and `country` carry the
   large effects, and Pearson correlation cannot see them at all. A correlation heatmap is the wrong
   instrument for this dataset — which is why §2.3–2.4 lead the analysis and this section supports it.
2. **Relationships are non-linear.** `lead_time` shows Pearson **0.184** but Spearman **0.225**; the
   binned view in §2.3 shows a clean 6.8x monotone gradient. `previous_cancellations` is the extreme
   case — Pearson **0.051** versus Spearman **0.127**, a 2.5x difference, because a single extreme
   value of 26 drags the linear fit while the rank correlation reads the real pattern. When Spearman
   materially exceeds Pearson, trust the binned rates.
3. **Correlation with a binary target is bounded by class balance**, so at 27.7% positives even a
   strong effect cannot reach a high r.

**Ranked by Spearman**, the honest ordering of numeric predictors is: `lead_time` (+0.225),
`required_car_parking_spaces` (−0.188, *leakage*), `adr` (+0.138), `total_of_special_requests`
(−0.129), `previous_cancellations` (+0.127), `booking_changes` (−0.124).

**Multicollinearity** is mild and only two pairs exceed |0.5|: `stays_in_week_nights` /
`stays_in_weekend_nights` (0.547) and `arrival_date_year` / `arrival_date_week_number` (−0.514, an
artifact of the incomplete 2015/2017 coverage). The engineered `total_nights` in Step 3 replaces the
first pair; note it correlates 0.95 with `stays_in_week_nights`, so use one or the other, never both.

**`days_in_waiting_list` is effectively uncorrelated** (+0.015) despite the 34.2% vs 27.4% gap noted
in §1.6 — because 99% of the column is zero, the binary `was_waitlisted` flag built in Step 3 is the
usable form.

| 21 | `booking_source_group` | categorical | Collapses segment x channel into 6 commercial channels |
| 22 | `is_group_block`, `is_child_only_record`, `is_zero_rated` | binary | Integrity flags from Step 1 — keep data, enable exclusion |
| 23 | `risk_score`, `cancellation_risk_category` | numeric/ordinal | Composite, leakage-free early-warning score |

### 3.1 Stay and guest features

```python
d = data_clean   # short alias; d and data_clean are the same object

d['total_nights']  = d['stays_in_week_nights'] + d['stays_in_weekend_nights']
d['total_guests']  = d['adults'] + d['children'] + d['babies']
d['weekend_ratio'] = (d['stays_in_weekend_nights'] / d['total_nights']).round(3)
d['is_family']     = ((d['children'] + d['babies']) > 0).astype(int)
d['adr_per_guest'] = (d['adr'] / d['total_guests'].replace(0, np.nan)).round(2)
d['total_revenue'] = (d['adr'] * d['total_nights']).round(2)


def guest_composition(r):
    '''Business-readable party type, ordered so the structural cases are caught first.'''
    kids = r['children'] + r['babies']
    if r['adults'] >= 5:  return 'Large party (5+)'
    if r['adults'] == 0:  return 'Child-only record'
    if kids > 0:          return 'Family with children'
    if r['adults'] == 1:  return 'Solo'
    if r['adults'] == 2:  return 'Couple'
    return 'Multi-adult party'


d['guest_composition']    = d.apply(guest_composition, axis=1)
d['stay_length_category'] = pd.cut(d['total_nights'], bins=[0, 1, 3, 7, 14, 100],
                                   labels=['1 night','2-3 nights','4-7 nights','1-2 weeks','Extended stay'])

print(d[['total_nights','total_guests','weekend_ratio','is_family','adr_per_guest',
         'total_revenue','guest_composition','stay_length_category']].head(8).to_string())

print("\nGuest composition mix:")
print(pd.DataFrame({'bookings': d.guest_composition.value_counts(),
                    'percent': (100 * d.guest_composition.value_counts() / len(d)).round(2),
                    'cancel_%': (100 * d.groupby('guest_composition')['is_canceled'].mean()).round(2),
                    'mean_adr': d.groupby('guest_composition')['adr'].mean().round(2)}).to_string())
```

**Business value.** `guest_composition` immediately separates the portfolio by both value and risk:
**Families** pay the highest rate (**ADR 157**) but cancel at **34.3%**; **Solo** travellers are the
cheapest (**80**) and by far the most reliable (**19.4%**); **Couples** are 65% of all bookings at
28.4%. That is a clean segmentation for targeted deposit and upsell policy, and none of it is visible
from `adults`/`children`/`babies` separately.

`total_revenue` is what makes §4.1 possible — cancellation *rate* treats a €75 one-night stay and a
€3,000 two-week suite booking as equal losses, which they are not.

### 3.2 Temporal features

```python
d['arrival_month_num'] = d['arrival_date_month'].map(MONTH_NUM)
d['arrival_date'] = pd.to_datetime(dict(year=d['arrival_date_year'],
                                        month=d['arrival_month_num'],
                                        day=d['arrival_date_day_of_month']))

print("Invalid arrival dates constructed:", int(d['arrival_date'].isna().sum()))
print("Arrival window: {} to {}".format(d['arrival_date'].min().date(), d['arrival_date'].max().date()))

d['arrival_quarter']     = 'Q' + d['arrival_date'].dt.quarter.astype(str)
season_map = {12: 'Winter', 1: 'Winter', 2: 'Winter', 3: 'Spring', 4: 'Spring', 5: 'Spring',
              6: 'Summer', 7: 'Summer', 8: 'Summer', 9: 'Autumn', 10: 'Autumn', 11: 'Autumn'}
d['booking_season']      = d['arrival_month_num'].map(season_map)
d['arrival_day_of_week'] = d['arrival_date'].dt.day_name()
d['is_weekend_arrival']  = d['arrival_date'].dt.dayofweek.isin([4, 5]).astype(int)

# high season defined from the data, not assumed
peak = d.groupby('arrival_month_num')['adr'].mean()
peak_months = peak[peak > peak.mean()].index.tolist()
d['is_high_season'] = d['arrival_month_num'].isin(peak_months).astype(int)
print("\nAbove-average-ADR months treated as high season:",
      [MONTH_ORDER[m - 1] for m in peak_months])

print("\nSeasonal performance:")
print(pd.DataFrame({
    'bookings': d.booking_season.value_counts(),
    'cancel_%': (100 * d.groupby('booking_season')['is_canceled'].mean()).round(2),
    'mean_adr': d.groupby('booking_season')['adr'].mean().round(2),
    'mean_revenue': d.groupby('booking_season')['total_revenue'].mean().round(2),
}).reindex(['Winter','Spring','Summer','Autumn']).to_string())
```

**Business value.** Building `arrival_date` doubles as a validation step: **zero invalid dates** were
produced, which confirms the three arrival-date part-columns are internally consistent (no 30 February
records) and that the window is 2015-07-01 to 2017-08-31.

`is_high_season` is derived from the data (the six above-average-ADR months, April–September) rather
than assumed from a calendar. The seasonal table then quantifies the exposure precisely: **Summer**
carries **31.7%** cancellation *and* **574** mean revenue per booking, versus **Autumn** at **23.5%**
and **322**. Peak season concentrates the most money in the least reliable bookings — which is the
economic argument for a seasonal deposit policy rather than a flat one.

### 3.3 Behavioural and commercial features

```python
d['total_previous_bookings'] = d['previous_cancellations'] + d['previous_bookings_not_canceled']
d['has_booking_history']     = (d['total_previous_bookings'] > 0).astype(int)
d['previous_cancel_ratio']   = (d['previous_cancellations']
                                / d['total_previous_bookings'].replace(0, np.nan)).fillna(0).round(3)
d['is_repeat_customer']      = ((d['is_repeated_guest'] == 1) | (d['previous_bookings_not_canceled'] > 0)).astype(int)
d['booking_modified']        = (d['booking_changes'] > 0).astype(int)
d['was_waitlisted']          = (d['days_in_waiting_list'] > 0).astype(int)

d['requests_category']         = pd.cut(d['total_of_special_requests'], bins=[-1, 0, 1, 2, 10],
                                        labels=['None','Low (1)','Medium (2)','High (3+)'])
d['booking_advance_category']  = d['lead_time_bin']
d['adr_category']              = pd.qcut(d['adr'], q=4, labels=['Budget','Standard','Premium','Luxury'])
d['is_domestic']               = (d['country'] == 'PRT').astype(int)


def booking_source_group(r):
    '''Collapse market_segment x distribution_channel into six commercial channels.'''
    ms, dc = r['market_segment'], r['distribution_channel']
    if ms == 'Online TA':                                    return 'Online OTA'
    if ms == 'Offline TA/TO':                                return 'Offline Travel Trade'
    if ms == 'Direct' or dc == 'Direct':                     return 'Direct'
    if ms in ('Corporate','Aviation') or dc == 'Corporate':  return 'Corporate / Business'
    if ms == 'Groups':                                       return 'Groups / Wholesale'
    return 'Complementary / Other'


d['booking_source_group'] = d.apply(booking_source_group, axis=1)

print(pd.DataFrame({
    'bookings': d.booking_source_group.value_counts(),
    'share_%': (100 * d.booking_source_group.value_counts() / len(d)).round(2),
    'cancel_%': (100 * d.groupby('booking_source_group')['is_canceled'].mean()).round(2),
    'mean_adr': d.groupby('booking_source_group')['adr'].mean().round(2),
    'revenue_share_%': (100 * d.groupby('booking_source_group')['total_revenue'].sum()
                        / d['total_revenue'].sum()).round(2),
}).sort_values('bookings', ascending=False).to_string())

print("\nADR quartile bands:")
print(d.groupby('adr_category', observed=True)['adr'].agg(['min','max','mean','size']).round(2).to_string())
```

**Business value.** `booking_source_group` turns two overlapping columns (8 segments x 5 channels =
40 sparse combinations) into six channels that a commercial team actually manages. The result is the
clearest strategic table in the notebook: **Online OTA is 59.2% of bookings, 64.6% of revenue, and
cancels at 35.6%** — the business's largest revenue source is also its least reliable.

`previous_cancel_ratio` fixes the problem diagnosed in §2.5: the raw `previous_cancellations` count is
so skewed (34.2) that its Pearson correlation collapses to 0.051. A *ratio* is bounded in $[0, 1]$
and comparable between a guest with 1 prior booking and one with 72.

These preserve the Step 1 findings as columns rather than deletions, so any later analysis can exclude
structurally atypical records without re-deriving them.

```python
d['is_group_block']       = ((d['adults'] >= 10) & (d['customer_type'] == 'Group')).astype(int)
d['is_child_only_record'] = ((d['adults'] == 0) & ((d['children'] + d['babies']) > 0)).astype(int)
d['is_zero_rated']        = (d['adr'] == 0).astype(int)

flags = ['is_group_block','is_child_only_record','is_zero_rated']
print(pd.DataFrame({
    'flagged_rows': [int(d[f].sum()) for f in flags],
    'share_%': [round(100 * d[f].mean(), 3) for f in flags],
    'cancel_%': [round(100 * d.loc[d[f] == 1, 'is_canceled'].mean(), 2) for f in flags],
}, index=flags).to_string())

print("\nThese flags are not filters - they let any downstream analysis exclude")
print("structurally atypical records on demand without deleting them now.")
```

### 3.5 Composite cancellation risk score

The final feature combines the independent signals found in Step 2 into one additive score. Three
design constraints:

1. **Leakage-free** — built only from fields known at booking time. No `reservation_status`, no
   `required_car_parking_spaces`, no `room_changed`.
2. **Weighted by observed effect size** — the three strongest findings (`Non Refund` at 94.7%, prior
   cancellations at 68.2%, and the lead-time gradient) get the heaviest weights; protective signals
   subtract.
3. **Interpretable** — a front-desk manager can read why a booking scored what it did, which a
   gradient-boosted model would not offer.

This is a transparent scorecard, not a fitted model: the weights are assigned from the effect sizes
measured above, so the validation below is a *consistency check* that the signals combine sensibly —
not an out-of-sample performance claim.

```python
def cancellation_risk_score(df):
    '''Additive, leakage-free risk score built only from information known at booking time.'''
    s = pd.Series(0, index=df.index, dtype=int)
    # horizon: the 6.8x monotone gradient from section 2.3
    s += pd.cut(df['lead_time'], bins=[-1, 7, 30, 90, 180, 10_000],
                labels=[0, 1, 2, 3, 4]).astype(int)
    # strongest categorical effects
    s += (df['deposit_type'] == 'Non Refund').astype(int) * 3
    s += (df['previous_cancellations'] >= 1).astype(int) * 3
    s += (df['market_segment'] == 'Online TA').astype(int)
    s += df['is_domestic']
    s += (df['customer_type'] == 'Transient').astype(int)
    # engagement signals
    s += (df['total_of_special_requests'] == 0).astype(int)
    s -= (df['total_of_special_requests'] >= 2).astype(int)
    s -= df['booking_modified']
    s -= df['is_repeat_customer']
    return s.clip(lower=0)


d['risk_score'] = cancellation_risk_score(d)
d['cancellation_risk_category'] = pd.cut(d['risk_score'], bins=[-1, 1, 3, 5, 99],
                                         labels=['Low','Moderate','High','Very High'])

tiers = pd.DataFrame({
    'bookings': d.cancellation_risk_category.value_counts(),
    'share_%': (100 * d.cancellation_risk_category.value_counts() / len(d)).round(2),
    'cancel_%': (100 * d.groupby('cancellation_risk_category', observed=True)['is_canceled'].mean()).round(2),
    'revenue_at_risk': d.groupby('cancellation_risk_category', observed=True)['total_revenue'].sum().round(0),
}).reindex(['Low','Moderate','High','Very High'])
print(tiers.to_string())

print("\nCancellation rate increases monotonically across tiers:", tiers['cancel_%'].is_monotonic_increasing)
print("Lift, Very High vs Low: {:.1f}x".format(tiers.loc['Very High','cancel_%'] / tiers.loc['Low','cancel_%']))

fig, ax = plt.subplots(1, 2, figsize=(14, 4.2))
bar_with_labels(ax[0], tiers.index, tiers['cancel_%'].values, color=[TEAL, GOLD, "#E98A5F", CORAL],
                fmt="{:.1f}%", ylabel="Cancellation rate (%)",
                title="Risk tier separates outcomes cleanly", baseline=BASE)
sc = d.groupby('risk_score').agg(n=('is_canceled','size'), c=('is_canceled','mean'))
sc = sc[sc.n >= 50]
ax[1].plot(sc.index, 100 * sc.c.values, marker='o', color=NAVY, linewidth=1.8, markersize=5)
ax[1].set_title("Cancellation rate by raw risk score (buckets with n>=50)")
ax[1].set_xlabel("Risk score"); ax[1].set_ylabel("Cancellation rate (%)")
ax[1].axhline(BASE, ls='--', lw=1.2, color=CORAL)
plt.tight_layout(); plt.show()
```

**Validation.** The tiers separate cleanly and monotonically:

| Tier | Bookings | Share | Cancellation rate |
|---|---|---|---|
| Low | 7,043 | 8.1% | **3.6%** |
| Moderate | 31,265 | 36.1% | **12.6%** |
| High | 35,052 | 40.5% | **31.5%** |
| Very High | 13,278 | 15.3% | **66.0%** |

An **18.2x lift** from the Low to the Very High tier, and the per-score curve rises smoothly with no
inversions — the component signals reinforce rather than cancel each other.

**Operational value.** The `Very High` tier is 15.3% of bookings but contains **€6.8m of booked
revenue at a 66% failure rate**. That is a small enough book to manage by hand — deposit requirements,
reconfirmation calls, or deliberate overbooking against it — without touching the 44% of bookings in
the Low/Moderate tiers that convert reliably.

**Rejected features** — worth recording, because knowing what *not* to build is part of the job:

* **`is_cancelled_before`** from `previous_cancellations > 0` — already covered by
  `previous_cancel_ratio` and `has_booking_history`; a third encoding of the same column adds
  collinearity, not information.
* **`revenue_lost`** (`total_revenue` where cancelled) — a direct function of the target. It belongs in
  reporting (§4.1), never in a feature set.
* **`room_upgraded` / `room_downgraded`** — tempting from the §2.3 heatmap, but room type letters are
  anonymised with no published quality ordering, so "upgrade" cannot be defined. `room_changed` is kept
  for description only, and is on the exclusion list anyway.
* **One-hot `country`, `agent`, `company`** — 862 combined levels would produce a sparse matrix wider
  than it is tall. `is_domestic` captures the effect that actually matters (§2.4); the rest should be
  frequency- or target-encoded if ever needed.
* **Any transformation of `adr`** — §1.8 showed `log1p` makes the distribution materially worse.

**The exclusion list is the most important output of this section.** `reservation_status` (a perfect
restatement of the target), `required_car_parking_spaces` (0 cancellations in 7,290 bookings) and
`room_changed` (assigned at check-in) all encode the outcome. A model trained with them would score
near-perfectly in validation and fail completely in production, because at booking time none of them
exists yet.

# Step 4 — Business Insights

## 4.1 Revenue at risk — reframing the cancellation problem

Cancellation *rate* is the wrong headline metric on its own, because it weights a €75 one-night stay
identically to a €3,000 two-week suite booking. The commercially meaningful question is how much money
is attached to bookings that fail.

```python
total_rev   = d['total_revenue'].sum()
lost_rev    = d.loc[d.is_canceled == 1, 'total_revenue'].sum()
nonref_lost = d.loc[(d.is_canceled == 1) & (d.deposit_type == 'Non Refund'), 'total_revenue'].sum()

print("Total booked room revenue     : {:>14,.0f}".format(total_rev))
print("Attached to cancelled bookings: {:>14,.0f}  ({:.1f}%)".format(lost_rev, 100 * lost_rev / total_rev))
print("  of which non-refundable     : {:>14,.0f}  (retained, not lost)".format(nonref_lost))
print("Net revenue genuinely at risk : {:>14,.0f}  ({:.1f}%)".format(
    lost_rev - nonref_lost, 100 * (lost_rev - nonref_lost) / total_rev))

seg = d.groupby('booking_source_group').agg(
    bookings=('is_canceled','size'), cancel_rate=('is_canceled','mean'),
    booked_revenue=('total_revenue','sum'), mean_adr=('adr','mean'))
seg['realised_revenue'] = d[d.is_canceled == 0].groupby('booking_source_group')['total_revenue'].sum()
seg['lost_revenue']     = seg['booked_revenue'] - seg['realised_revenue']
seg['lost_%']           = (100 * seg['lost_revenue'] / seg['booked_revenue']).round(1)
seg['cancel_%']         = (100 * seg['cancel_rate']).round(2)
seg = seg.drop(columns='cancel_rate').sort_values('booked_revenue', ascending=False)
print("\n" + seg.round(0).to_string())

fig, ax = plt.subplots(1, 2, figsize=(15, 4.5))
srt = seg.sort_values('lost_revenue', ascending=True)
ax[0].barh(srt.index, srt['realised_revenue'] / 1e6, color=TEAL, label='Realised')
ax[0].barh(srt.index, srt['lost_revenue'] / 1e6, left=srt['realised_revenue'] / 1e6,
           color=CORAL, label='Lost to cancellation')
ax[0].set_xlabel("Room revenue (EUR millions)"); ax[0].set_title("Revenue realised vs lost by channel")
ax[0].legend(frameon=False, loc='lower right')
for y, (r, l) in enumerate(zip(srt['realised_revenue'], srt['lost_revenue'])):
    ax[0].text((r + l) / 1e6, y, f"  {100*l/(r+l):.0f}% lost", va='center', fontsize=8)
bar_with_labels(ax[1], seg.index, seg['cancel_%'].values, color=NAVY, fmt="{:.1f}%",
                rotate=30, ylabel="Cancellation rate (%)",
                title="Cancellation rate by channel", baseline=BASE)
plt.tight_layout(); plt.show()
```

#### Insight — €11.2m of revenue is genuinely exposed, and 83% of it sits in one channel

Of **€34.4m** in booked room revenue, **€11.5m (33.3%)** is attached to cancelled bookings. Two
corrections matter:

* **€237k of that is non-refundable and therefore retained**, not lost. Net exposure is **€11.24m
  (32.6%)**.
* **Revenue exposure exceeds booking exposure** — 33.3% of revenue against a 27.7% cancellation rate.
  Cancelled bookings are systematically *more valuable* than kept ones, because the high-ADR, long-stay,
  long-lead bookings are the ones most likely to fail.

**Online OTA alone accounts for €9.29m of the €11.24m — 83% of all exposure.** It is 59% of bookings,
65% of revenue, and loses **42%** of what it books. Every other channel loses 8–29%.

This reframes the strategic problem. It is not "reduce cancellations across the business"; it is
"the OTA channel is simultaneously the largest revenue source and the origin of five-sixths of the
revenue risk." Shifting even a tenth of OTA volume to Direct — which loses 19% instead of 42% —
would be worth roughly **€500k** in realised revenue, before any commission saving.

## 4.2 What actually drives cancellations

```python
def rate_gap(df, hi, lo):
    '''Cancellation rate for two query groups, and the gap between them in percentage points.'''
    a = df.loc[df.eval(hi), 'is_canceled'].mean() * 100
    b = df.loc[df.eval(lo), 'is_canceled'].mean() * 100
    return a, b, a - b


drivers = [
    ("Non-refundable deposit", "deposit_type == 'Non Refund'", "deposit_type == 'No Deposit'"),
    ("Prior cancellation on file", "previous_cancellations >= 1", "previous_cancellations == 0"),
    ("Domestic group booking", "is_domestic == 1 and market_segment == 'Groups'",
                               "is_domestic == 0 and market_segment == 'Groups'"),
    ("Lead time > 180 days", "lead_time > 180", "lead_time <= 7"),
    ("Online OTA channel", "booking_source_group == 'Online OTA'", "booking_source_group == 'Direct'"),
    ("No special requests", "total_of_special_requests == 0", "total_of_special_requests >= 2"),
    ("Domestic guest", "is_domestic == 1", "is_domestic == 0"),
    ("Transient customer", "customer_type == 'Transient'", "customer_type == 'Transient-Party'"),
    ("City hotel", "hotel == 'City Hotel'", "hotel == 'Resort Hotel'"),
    ("Never amended booking", "booking_changes == 0", "booking_changes >= 1"),
    ("First-time guest", "is_repeat_customer == 0", "is_repeat_customer == 1"),
]

rows = []
for name, hi, lo in drivers:
    a, b, gap = rate_gap(d, hi, lo)
    rows.append({"driver": name, "high_risk_group_%": round(a, 1),
                 "comparison_group_%": round(b, 1), "gap_pp": round(gap, 1)})
drv = pd.DataFrame(rows).sort_values("gap_pp", ascending=False).reset_index(drop=True)
print(drv.to_string(index=False))

fig, ax = plt.subplots(figsize=(9, 5))
srt = drv.sort_values("gap_pp")
bars = ax.barh(srt['driver'], srt['gap_pp'], color=[CORAL if v > 0 else TEAL for v in srt['gap_pp']],
               edgecolor='white', linewidth=.5)
ax.bar_label(bars, labels=[f"+{v:.1f} pp" for v in srt['gap_pp']], padding=3, fontsize=8.5)
ax.set_xlabel("Cancellation-rate gap vs comparison group (percentage points)")
ax.set_title("What actually drives cancellations", fontweight='bold')
ax.set_xlim(0, srt['gap_pp'].max() * 1.22)
plt.tight_layout(); plt.show()
```

#### Insight — ranked drivers, with the causal caveat attached

| Rank | Driver | High-risk | Comparison | Gap |
|---|---|---|---|---|
| 1 | Non-refundable deposit | 94.7% | 26.9% | **+67.8 pp** |
| 2 | Domestic group booking | 58.8% | 1.6% | **+57.1 pp** |
| 3 | Prior cancellation on file | 68.2% | 26.9% | **+41.3 pp** |
| 4 | Lead time > 180 days | 39.8% | 8.5% | **+31.3 pp** |
| 5 | First-time guest | 28.6% | 7.6% | **+21.1 pp** |
| 6 | Online OTA channel | 35.5% | 14.8% | **+20.7 pp** |

**The ranking must be read with direction of causation in mind, which differs by row:**

* **#1 is not causal.** As established in §2.3, `Non Refund` terms are *applied to* bookings already
  judged risky. Requiring deposits more widely would not reproduce this gap.
* **#2, #3, #4 are genuine, actionable risk markers** — guest origin, behavioural history and booking
  horizon are all known at booking time and are not consequences of the outcome.
* **#5 and #6 are structural** — they describe which book of business is fragile, not a lever to pull
  directly.

**The most useful finding is #2**, because it is both the second-largest effect and precisely
targetable: domestic Groups cancel at **58.8%** against **1.6%** for the same segment sold
internationally, a **37x** difference. Combined with domestic Offline TA/TO (**44.4%** vs **0.5%**),
this is roughly **6,700 bookings** where intervention is justified — and where it would not disturb
the reliable international base.

## 4.3 Answers to the core business questions

**Which customers cancel the most?**
Domestic (Portuguese) guests booking through Groups (**58.8%**) or Offline TA/TO (**44.4%**) channels,
and any guest with a prior cancellation on file (**68.2%**). Overall, domestic guests cancel at
**36.4%** versus **23.8%** international. By party type, families cancel at **34.3%** and solo
travellers at **19.4%**.

**Which booking channels are the most reliable?**
`Corporate / Business` (**12.9%**), `Direct` (**14.8%**) and `Offline Travel Trade` (**15.0%**) — but
the last only *internationally*; its domestic book runs at 44.4%. `Online OTA` at **35.6%** is the
least reliable and the largest.

**Which customer types generate the highest ADR?**
`Transient` at **111** mean ADR, ahead of `Contract` (93), `Transient-Party` (88) and `Group` (86).
Crossed with channel, `Online TA` + `Transient` is the top-yielding cell (**119**). The commercial
tension is that the highest-ADR segments are also the highest-cancelling ones.

**Which hotel performs better?**
Neither, unqualified — they are different businesses (§2.3). The City Hotel has higher headline ADR
(**111.66** vs **100.12**), 61% of volume and steady year-round demand, but cancels more (**30.2%** vs
**23.7%**). The Resort has extreme seasonality (ADR **49.92** in January to **188.90** in August),
cancels less, and earns **more per realised booking** (**422** vs **328**). The City Hotel is a
volume-and-occupancy business; the Resort is a yield-management business.

**Which countries have the highest cancellation rate?**
Among countries with 500+ bookings: **China (CHN) 48.1%**, **Russia 39.9%**, **Brazil 36.5%**,
**Portugal 36.4%**, **Italy 35.1%**. The most reliable are **Austria 17.9%**, **Netherlands 18.3%**,
**Great Britain 19.1%**, **Germany 19.6%** and **France 19.6%**. Portugal matters most because it is
31% of all bookings — the others are individually small. (Note the `CN`/`CHN` coding issue flagged in
§2.2 before acting on the China figure.)

**Which market segments are the most valuable?**
On revenue contribution, `Online OTA` is unmatched (**64.6%** of booked revenue) but loses 42% of it.
On *realised* revenue per booking, `Direct` is the strongest combination: ADR **111** with only
**14.8%** cancellation. `Corporate` is the most reliable (12.9%) but cheap (ADR **70**) and small
(5.7%). There is no segment that is simultaneously large, high-rate and reliable.

**Which factors most influence cancellations?**
In order of measured effect: prior cancellation history, guest origin interacted with channel, booking
lead time, channel, and engagement signals (special requests, booking amendments). Deposit type shows
the largest raw gap but is a *consequence* of risk assessment rather than a cause.

## 4.4 Recommendations

**1. Fix the domestic Groups / Offline TA/TO leak first.** ~6,700 bookings cancelling at 44–59%
against international rates of 0.5–1.6% on the identical product. Require a partial deposit or a
72-hour reconfirmation for domestic group and offline-trade bookings only. This is the highest
effect-per-unit-of-disruption action available and it leaves the reliable international base untouched.

**2. Treat the `Very High` risk tier as a managed book.** 13,278 bookings, 66% cancellation, €6.8m
attached. Small enough for active management — reconfirmation contact, deposit terms, or planned
overbooking against it — and the scorecard is transparent enough for front-desk staff to act on.

**3. Shift OTA volume to Direct.** Direct loses 19% of booked revenue versus 42% for OTA, at a
comparable ADR (111 vs 119). Moving 10% of OTA volume to Direct is worth roughly €500k in realised
revenue before commission savings.

**4. Solicit preferences from long-lead bookers.** Special requests cut cancellation from 44.5% to
24.0% at a 91–180 day horizon, independent of lead time (§2.4). A pre-arrival preference email is
close to free and targets the exact population that fails most.

**5. Make deposit policy seasonal, not flat.** Summer carries both the highest cancellation rate
(31.7%) and the highest revenue per booking (574). A flat 98.7% no-deposit policy under-protects
exactly the inventory that is worth most.

**6. Re-price or restructure long stays.** The 15–30 night band cancels at over **54%** while occupying
inventory for weeks at a discounted rate. Either a deposit or a shorter free-cancellation window is
warranted.

**7. Resolve two data-quality items before they distort reporting.** The `CN`/`CHN` country coding
split (§2.2) and the Resort-only `meal = Undefined` category need a source-system answer from the
business.

## Limitations

Stated plainly, because they bound every number above:

* **The date window is not three full years.** 2015 covers July–December and 2017 January–August; only
  2016 is complete. All year-over-year comparisons must be month-matched, and no genuine trend can be
  read from three partial years.
* **31,994 exact duplicates were removed** on the assumption that identical rows are duplicate records.
  This dataset has no booking ID, so it is possible some are genuinely distinct same-day bookings of
  the same room type by a group. The cancellation rate is 27.7% de-duplicated versus 37.0% raw, so this
  assumption materially affects every rate in the notebook and should be confirmed against the source
  system.
* **`agent` and `company` remain high-cardinality identifiers** (335 and 349 levels) and were not
  encoded; the `Unknown Agent` / `Unknown Company` buckets from the imputation step mix genuinely
  absent values with unrecorded ones.
* **The risk score is a hand-weighted scorecard, not a fitted model.** Its weights come from effect
  sizes measured on this same data, so the 18.2x lift is an in-sample consistency check, not an
  out-of-sample performance estimate. A hold-out validation is required before operational use.
* **All findings are associational.** The deposit finding (§2.3) is the clearest illustration of why no
  row in the driver table should be read as causal without a controlled test.
